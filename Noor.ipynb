```python
# Core 1: Product sales dictionary
products = {
    "Laptop": 25,
    "Mouse": 80,
    "Keyboard": 45,
    "Monitor": 30,
    "Headphones": 60
}

total_units = 0

for product, units in products.items():
    print(product, ":", units, "units")
    total_units = total_units + units

print("Total units of all products:", total_units)
```

```python
# Core 2: Count months with sales above 50
sales = [40, 55, 62, 48, 70, 35, 90, 52, 45, 60, 58, 41]

count = 0
for s in sales:
    if s > 50:
        count = count + 1

print("Months with sales above 50:", count)
```

```python
# Core 3: Grade from marks
marks = int(input("Enter your marks (0-100): "))

if marks >= 80:
    print("Grade: A")
elif marks >= 70:
    print("Grade: B")
elif marks >= 50:
    print("Grade: C")
else:
    print("Grade: F (Fail)")
```

```python
# Stretch 1: Pandas DataFrame
import pandas as pd

df = pd.DataFrame({
    "Month": ["Jan", "Feb", "Mar", "Apr", "May", "Jun",
              "Jul", "Aug", "Sep", "Oct", "Nov", "Dec"],
    "Sales": [40, 55, 62, 48, 70, 35, 90, 52, 45, 60, 58, 41]
})

print(df)
print("Total Sales  :", df["Sales"].sum())
print("Average Sales:", df["Sales"].mean())
print("Maximum Sales:", df["Sales"].max())
```

```python
# Stretch 2: Bar chart
import matplotlib.pyplot as plt

df.plot(x="Month", y="Sales", kind="bar")
plt.title("Monthly Sales")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.show()
```
